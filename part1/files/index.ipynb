## ⚠️  First time here? Run the pre-download step now

Before the tutorial session, open any notebook below, run its **first cell** (the setup cell), and wait for _Setup complete_.
This downloads ~240 MB of data **once** and caches it in your browser.  
On the day it will load in seconds instead of minutes.

> Use the same browser and same profile on the day. Do **not** use incognito/private mode.
